# 20a supervised market-weather feature matrix

This notebook is a thin wrapper around the Terminal-ready 20a script. The script constructs leakage-controlled weather-only, market-only and combined feature sets, assigns one date group to every HKO settlement date, and writes a review bundle.


```python
from pathlib import Path
import subprocess
import sys

cwd = Path.cwd().resolve()
repo_root = cwd.parent if cwd.name == 'notebooks' else cwd
script = repo_root / 'scripts' / '20a_supervised_feature_matrix.py'
cmd = [sys.executable, str(script), '--repo-root', str(repo_root)]
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

```python
import json
import pandas as pd

processed = repo_root / 'data' / 'processed'
matrix = pd.read_csv(processed / '20a_supervised_feature_matrix.csv')
dictionary = pd.read_csv(processed / '20a_feature_dictionary.csv')
checks = pd.read_csv(processed / '20a_integrity_checks.csv')
feature_sets = json.loads((processed / '20a_model_feature_sets.json').read_text())

print('Matrix shape:', matrix.shape)
print('Feature-set sizes:', {k: len(v) for k, v in feature_sets.items()})
display(checks)
display(dictionary.head(20))
```

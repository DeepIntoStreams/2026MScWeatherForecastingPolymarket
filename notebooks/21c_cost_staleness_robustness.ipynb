# 21c transaction costs, staleness and robustness — final locked-holdout version

This version computes empirical staleness quantiles from exactly 440 locked-holdout snapshots: 10 dates × 4 decision rules × 11 contracts.


```python
from pathlib import Path
import subprocess, sys
cwd = Path.cwd().resolve()
repo_root = cwd.parent if cwd.name == 'notebooks' else cwd
script = repo_root / 'scripts' / '21c_cost_staleness_robustness.py'
cmd = [sys.executable, str(script), '--repo-root', str(repo_root), '--cost-grid', '0,0.005,0.01,0.02,0.05', '--staleness-quantiles', '0.25,0.50,0.75,0.90', '--threshold-grid', '0.02,0.05,0.10', '--primary-cost', '0.01', '--primary-staleness-quantile', '0.50', '--primary-threshold', '0.05']
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

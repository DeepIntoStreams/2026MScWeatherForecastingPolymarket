# 18b HKO certified upper-tail outcome and decision panel

This notebook converts the certified Hong Kong upper-tail contract universe and the 18a CLOB price-history panel into a conservative no-lookahead decision panel.

The primary decision rule is the last observed YES price before 00:00 Hong Kong time on the event date. Additional 12-hour and 24-hour prior rules are included as robustness checks.

This is still a market-implied probability panel, not a fully executable trading backtest.

```python
from pathlib import Path
import runpy


def find_repo_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for p in [start, *start.parents]:
        if (p / '.git').exists():
            return p
    return start

repo = find_repo_root()
script = repo / 'scripts' / '18b_hko_certified_upper_tail_outcome_and_decision_panel.py'
print('Repository root:', repo)
print('Running script:', script)
runpy.run_path(str(script), run_name='__main__')
```

## Inspect key outputs

```python
import pandas as pd
from pathlib import Path

repo = find_repo_root()
decision_path = repo / 'data' / 'processed' / '18b_hko_certified_upper_tail_decision_panel.csv'
scoring_path = repo / 'data' / 'processed' / '18b_hko_certified_upper_tail_scoring_ready_market_panel.csv'
outcome_path = repo / 'data' / 'processed' / '18b_hko_certified_upper_tail_outcome_status.csv'

decision = pd.read_csv(decision_path)
scoring_ready = pd.read_csv(scoring_path)
outcomes = pd.read_csv(outcome_path)

print('Decision panel shape:', decision.shape)
print('Scoring-ready shape:', scoring_ready.shape)
print('Outcome status shape:', outcomes.shape)

print('\nDecision status counts:')
print(decision['decision_panel_status'].value_counts(dropna=False))

print('\nOutcome availability counts:')
print(outcomes['outcome_available'].value_counts(dropna=False))

print('\nDecision panel preview:')
preview_cols = [c for c in [
    'event_date', 'threshold_K', 'decision_rule', 'decision_timestamp_utc',
    'yes_price', 'hko_tmax_C', 'Y_ge_K', 'outcome_available',
    'decision_panel_status', 'price_staleness_hours'
] if c in decision.columns]
print(decision[preview_cols].to_string(index=False))
```

# 19c ECMWF proxy bias and scale correction

This notebook is a thin review wrapper around the Terminal-ready 19c script. The script performs leakage-free expanding-window bias and residual-scale correction, regenerates HKO event probabilities, evaluates exact common support, and writes a review bundle.


```python
from pathlib import Path
import subprocess
import sys

cwd = Path.cwd().resolve()
repo_root = cwd.parent if cwd.name == 'notebooks' else cwd
script = repo_root / 'scripts' / '19c_ecmwf_proxy_bias_scale_calibration.py'

cmd = [
    sys.executable,
    str(script),
    '--repo-root', str(repo_root),
    '--min-train-dates', '14',
    '--outcome-availability-lag-hours', '24',
    '--bootstrap-reps', '5000',
]
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

```python
import pandas as pd

processed = repo_root / 'data' / 'processed'
binary = pd.read_csv(processed / '19c_binary_score_summary.csv')
temperature = pd.read_csv(processed / '19c_temperature_error_summary.csv')
categorical = pd.read_csv(processed / '19c_categorical_score_summary.csv')
checks = pd.read_csv(processed / '19c_integrity_checks.csv')

display(binary)
display(temperature)
display(categorical)
display(checks)
```

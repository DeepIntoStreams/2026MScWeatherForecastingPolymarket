# 21b full Hong Kong event-book trading simulation

This notebook wraps the Terminal-ready 21b script. It treats each complete HKO contract book as a mutually exclusive portfolio while preserving the frozen 20e forecasts.


```python
from pathlib import Path
import subprocess, sys
cwd = Path.cwd().resolve()
repo_root = cwd.parent if cwd.name == 'notebooks' else cwd
script = repo_root / 'scripts' / '21b_full_event_book_trading.py'
cmd = [sys.executable, str(script), '--repo-root', str(repo_root), '--thresholds', '0,0.02,0.05,0.10', '--primary-threshold', '0.05', '--trade-cost-per-position', '0']
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

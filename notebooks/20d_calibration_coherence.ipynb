# 20d out-of-fold calibration and event-book coherence — corrected version

This replaces the earlier 20d implementation. It uses a robust log-score-first calibration rule with an explicit extreme-probability guard and leaves the final chronological holdout sealed.


```python
from pathlib import Path
import subprocess, sys
cwd = Path.cwd().resolve()
repo_root = cwd.parent if cwd.name == 'notebooks' else cwd
script = repo_root / 'scripts' / '20d_calibration_coherence.py'
cmd = [sys.executable, str(script), '--repo-root', str(repo_root), '--min-calibration-dates', '8', '--bootstrap-reps', '5000', '--random-seed', '20260715']
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```

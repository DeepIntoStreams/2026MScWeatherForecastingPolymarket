# Notebook 14: Final Empirical Synthesis

This notebook consolidates the locked empirical evidence from
Notebooks 04 to 13. It does not reselect the probabilistic model,
continuous calibration, event-probability calibration or trading
strategy.

The notebook records the empirical results together with the limits
on the conclusions that may be drawn from them.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve()

for candidate in (ROOT, *ROOT.parents):
    if (
        candidate
        / "config/final_empirical_synthesis_spec.yaml"
    ).exists():
        ROOT = candidate
        break
else:
    raise FileNotFoundError(
        "Repository root was not found."
    )

completed = subprocess.run(
    [
        sys.executable,
        str(
            ROOT
            / "tools/"
            "build_notebook14_final_synthesis.py"
        ),
    ],
    cwd=ROOT,
    text=True,
    capture_output=True,
)

print(completed.stdout)

if completed.returncode != 0:
    print(completed.stderr)
    raise RuntimeError(
        "Notebook 14 synthesis failed."
    )
```

# Notebook 15: Final Reproducibility and Release Audit

This notebook certifies the empirical lineage recorded through
Notebook 14. It verifies hashes, canonical notebook execution,
repository tests, branch synchronisation and the unchanged status
of all locked empirical choices.

The audited release is the parent empirical commit. Notebook 15
does not fit, recalibrate, reselect or evaluate a new model.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve()

for candidate in (ROOT, *ROOT.parents):
    if (
        candidate
        / "config/reproducibility_release_spec.yaml"
    ).exists():
        ROOT = candidate
        break
else:
    raise FileNotFoundError(
        "Repository root not found."
    )

completed = subprocess.run(
    [
        sys.executable,
        str(
            ROOT
            / "tools/"
            "build_notebook15_release_audit.py"
        ),
    ],
    cwd=ROOT,
    text=True,
    capture_output=True,
)

print(completed.stdout)

if completed.returncode != 0:
    print(completed.stderr)
    raise RuntimeError(
        "Notebook 15 release audit failed."
    )
```

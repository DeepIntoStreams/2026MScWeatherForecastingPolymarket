# Notebook 16: Thesis Evidence and Empirical Reporting

This notebook converts the certified empirical release into compact
thesis tables, LaTeX files, figures and a numerical results register.

No model, calibration parameter or trading strategy is selected or
changed in this notebook.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve()

for candidate in (ROOT, *ROOT.parents):
    if (
        candidate
        / "config/thesis_evidence_spec.yaml"
    ).exists():
        ROOT = candidate
        break
else:
    raise FileNotFoundError(
        "Repository root not found."
    )

completed = subprocess.run(
    [
        sys.executable,
        str(
            ROOT
            / "tools/"
            "build_notebook16_thesis_evidence.py"
        ),
    ],
    cwd=ROOT,
    text=True,
    capture_output=True,
)

print(completed.stdout)

if completed.returncode != 0:
    print(completed.stderr)
    raise RuntimeError(
        "Notebook 16 evidence extraction failed."
    )
```

## Locked model and calibration choices

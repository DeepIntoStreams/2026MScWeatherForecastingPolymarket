# 21e final Hong Kong empirical release freeze

This notebook wraps the 21e release-freeze script. It records the exact Git commit, hashes the final empirical artefacts, validates source integrity blocks, and writes a reproducible release manifest.


```python
from pathlib import Path
import subprocess, sys
cwd = Path.cwd().resolve()
repo_root = cwd.parent if cwd.name == 'notebooks' else cwd
script = repo_root / 'scripts' / '21e_freeze_final_empirical_release.py'
subprocess.run([sys.executable, str(script), '--repo-root', str(repo_root), '--release-name', 'hko_empirical_release_v1'], check=True)
```

```python
import json, pandas as pd
release_dir = repo_root / 'data' / 'releases' / 'hko_empirical_release_v1'
display(pd.read_csv(release_dir / '21e_release_integrity_checks.csv'))
display(pd.read_csv(release_dir / '21e_release_integrity_summary.csv'))
display(pd.read_csv(release_dir / '21e_release_file_index.csv').head(30))
print(json.loads((release_dir / '21e_release_manifest.json').read_text()))
```

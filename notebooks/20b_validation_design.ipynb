# 20b date-grouped validation design

This notebook is a thin wrapper around the Terminal-ready 20b script. It freezes expanding-window date-grouped cross-validation and the final chronological holdout before any model fitting.


```python
from pathlib import Path
import subprocess, sys

cwd = Path.cwd().resolve()
repo_root = cwd.parent if cwd.name == 'notebooks' else cwd
script = repo_root / 'scripts' / '20b_validation_design.py'
cmd = [
    sys.executable, str(script),
    '--repo-root', str(repo_root),
    '--holdout-fraction', '0.20',
    '--min-holdout-dates', '8',
    '--n-cv-folds', '4',
    '--min-train-dates', '16',
]
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```
